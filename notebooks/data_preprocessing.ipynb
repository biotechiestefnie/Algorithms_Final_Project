# Data Preparation Process for Prototype and Final Implementations with data_preparation.py

This implementation notebook serves as a record of all data preparation steps taken from sampling the raw datafiles to splitting datasets for the training and testing phases of the final execution. All raw data was retrieved from UCSC Genome Browser's and GENCODE's v19 *H. sapiens* GRCh37.p13.genome assembly, release and freeze year 2013. This was a mandatory design strategy to ensure all datasets were made of high-quality, experimentally validated sequences with extensive use histories.

The promoter sequences were obtained from the EPDnew v006 track as hosted within the UCSC Genome Browser for the hg19 assembly. Although EPDnew is curated by the Swiss Institute of Bioinformatics, the UCSC‑hosted version ensures that all promoter coordinates and extracted sequences are aligned to the same hg19/GRCh37 reference genome used for the rest of the dataset. Exons and introns were extracted directly from the GENCODE v19 annotation and the corresponding GRCh37.p13 reference genome, which is the Genome Reference Consortium assembly that UCSC designates as hg19. GENCODE v19 is the final GENCODE release for GRCh37, and no exon or intron FASTA files were ever distributed for this assembly. Additionally, UCSC deprecated hg19 feature‑level FASTA export around 2023, which required downloading the full GRCh37.p13 genome FASTA and reconstructing exon and intron sequences locally using gffread.
